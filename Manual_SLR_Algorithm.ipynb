```python
import numpy as np
```

```python
def cost(x, y, w, b):
    m = len(x)
    cost = 0
    for i in range(m):
        temp = (x[i] * w) + b - y[i]
        cost += temp**2
    cost /= 2*m
    return cost
```

```python
def gradient(x, y, w, b):
    w_temp = 0
    b_temp = 0
    m = len(x)

    for i in range(m):
        temp = ((x[i] * w) + b - y[i]) * x[i]
        w_temp += temp
    w_deriv = w_temp / m
    for i in range(m):
        temp = (x[i] * w) + b - y[i]
        b_temp += temp
    b_deriv = b_temp / m

    return w_deriv, b_deriv
```

```python
def gradient_descent(x, y, w_in, b_in, a, n):
    w = w_in
    b = b_in
    print(f"w: {round(w, 2)}, b: {round(b, 2)}, cost : {round(cost(x,y,w,b), 2)}")

    for i in range(n):
        w_deriv, b_deriv = gradient(x, y, w, b)

        w = w - a* w_deriv
        b = b - a* b_deriv

        if i%50==0:
            print(f"w: {round(w, 2)}, b: {round(b, 2)}, cost : {round(cost(x,y,w,b), 2)}")
    return w, b
```

```python
x = [1, 2, 3, 4, 5]
y = [100, 200, 300, 400, 500]

w, b = gradient_descent(x, y, 0, 4, 0.1, 1000)
```

```python
def predict(x):
    y = x*w + b
    return y
```
