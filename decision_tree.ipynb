```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, recall_score, confusion_matrix, r2_score
import matplotlib.pyplot as plt
```

Dataset:

```python
data = pd.read_csv("species.csv")
data.dropna(inplace=True)

X = data.drop(["Species", "Tree_ID"], axis=1)
X = pd.get_dummies(X, columns=['Leaf_Type', 'Bark_Texture'])
y = data['Species']
```

Train Test Val split:

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

Shape inspection: 

```python
print("Training: ",np.shape(X_train), np.shape(y_train))
print("Test data: ", np.shape(X_test), np.shape(y_test))
```

```python
model = DecisionTreeClassifier(
    max_depth=3,
    min_samples_split=10,
    min_samples_leaf=6
)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
accur = accuracy_score(y_test, y_pred)
accur
```

```python
plt.figure(figsize=(16, 10))

plot_tree(
    model,
    feature_names=X.columns,
    class_names=model.classes_,
    filled=True,
    rounded=True,
    fontsize=9
)

plt.show()
```
