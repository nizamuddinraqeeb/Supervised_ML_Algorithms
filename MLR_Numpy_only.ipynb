```python
import numpy as np
```

```python
def cost(X, y, w, b):
    m = X.shape[0]
    cost = 0

    for i in range(m):
        y_hat = np.dot(X[i], w) + b
        err = y_hat - y[i]
        cost += err**2
    cost /= (2*m)
    return cost
```

```python
def gradient(X, y, w, b):
    m, n = X.shape
    dw = np.zeros_like(X[0])
    db = 0

    for i in range(m):
        y_hat = np.dot(X[i], w) + b
        err = y_hat - y[i]
        for j in range(n):
            dw[j] += err*X[i, j]
        db += err
    dw /= m
    db /= m

    return dw, db
```

```python
def gradient_descent(X, y, w_in, b_in, a, n):
    w = w_in
    b = b_in

    for i in range(n):
        dw, db = gradient(X, y, w, b)
        w = w - (a*dw)
        b = b - (a*db)

        if i%50==0:
            print(f"{i}th iteration cost: {cost(X, y, w, b)}")
    return w, b
```

```python
X = np.array([[10, 10, 5],[20, 10, 0],[4, 5, 5],[5, 10, 5]]).astype(float)
y = np.array([25, 30, 14,20])
```

```python
w = np.zeros(X.shape[1])
b = 0
w, b = gradient_descent(X, y, w,b, 0.001, 10000)
print(f"Weights: {w}, Bias: {round(b, 2)}")
```

```python
def predict(x):
    y_hat = np.dot(x, w) +b
    return y_hat
```

```python
y_hat = predict(X)
```

```python
import matplotlib.pyplot as plt
plt.title("Predicted vs Actual Value")

plt.scatter(y_hat, y_hat, color='red', label="Predicted")
plt.scatter(y, y, color='green', label="Actual")
plt.legend()
plt.show()
```
